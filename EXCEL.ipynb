Question 1: Paste vs. Paste Special
Paste (Ctrl+V) copies all elements from the source, including values, formulas, formatting, and styles. Paste Special offers selective pasting options like values only, formats, or formulas, preventing unwanted changes such as overwriting formats.
​

For example, pasting a formula as values converts =A1+B1 to its result (e.g., 10), useful for sharing static data.
​

Question 2: Freeze Panes and Split Panes
Freeze Panes locks top rows and/or left columns in place during scrolling, ideal for keeping headers visible in large datasets.
​
​

Split Panes divides the window into adjustable sections for simultaneous viewing of distant areas, without locking, useful for comparing non-adjacent data.

Question 3: Insert Row vs. Column
Inserting a row adds horizontal space above the selected row(s), shifting lower rows down. Inserting a column adds vertical space left of the selected column(s), shifting right columns over.
​
​

Yes, select multiple contiguous rows/columns (e.g., drag or Shift+click), right-click > Insert, or use Ctrl+Shift++ for bulk insertion.

Question 4: Logical Functions
Logical functions evaluate conditions to return TRUE/FALSE or specific values, enabling decision-based calculations.
​

Examples: IF(A1>10, "High", "Low") checks a condition and outputs text; AND(A1>10, B1<5) returns TRUE only if both are true, used in validations.
​

Question 5: XLOOKUP vs. VLOOKUP
XLOOKUP searches a range or array for a match and returns corresponding data from another range, supporting bidirectional lookup, wildcards, and exact/approximate matches.
​

Unlike VLOOKUP, which scans left-to-right only from the first column and requires column index, XLOOKUP is more flexible, handles errors better (default #N/A), and doesn't need sorted data.
